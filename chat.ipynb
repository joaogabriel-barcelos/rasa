# 🤖 Chatbot da Plataforma de Ensino

Rode a célula abaixo (Shift+Enter) e depois **digite sua pergunta no campo de texto e aperte Enter**.

Sugestões de perguntas: *"Quais cursos tem disponíveis?"*, *"Quanto custa para estudar aqui?"*, *"Qual a diferença do acesso pago ao gratuito?"*.

```python
import asyncio
import nest_asyncio
nest_asyncio.apply()

from rasa.core.agent import Agent

agent = Agent.load("models")
print("Modelo carregado! Pode conversar na célula abaixo.")
```

```python
import ipywidgets as widgets
from IPython.display import display

output = widgets.Output()
entrada = widgets.Text(
    placeholder="Digite sua pergunta e aperte Enter...",
    layout=widgets.Layout(width="100%")
)

def enviar(_):
    pergunta = entrada.value
    entrada.value = ""
    if not pergunta.strip():
        return
    with output:
        print(f"Você: {pergunta}")
        respostas = asyncio.get_event_loop().run_until_complete(agent.handle_text(pergunta))
        for r in respostas:
            print(f"Bot: {r.get('text', '')}")
        print()

entrada.on_submit(enviar)
display(output, entrada)
```
